```python
from pyspark.sql import functions as F

# =====================================================
# CREATE SAMPLE DATAFRAMES
# =====================================================

data = [
    (1, "Rahul", 25, 50000, "IT"),
    (2, "Priya", 28, 60000, "HR"),
    (3, "Amit", 30, 70000, "IT"),
    (4, "Sneha", 24, 45000, "Finance"),
    (5, "Kiran", 35, 80000, "IT"),
    (6, "Priya", 28, 60000, "HR"),
    (7, "Vijay", None, None, "Sales")
]

columns = ["id", "name", "age", "salary", "department"]

df = spark.createDataFrame(data, columns)

dept_data = [
    ("IT", "Bengaluru"),
    ("HR", "Hyderabad"),
    ("Finance", "Mumbai"),
    ("Sales", "Chennai")
]

dept_df = spark.createDataFrame(
    dept_data, ["department", "location"]
)

print("ORIGINAL DATAFRAME")
df.show()

# 1. SELECT: Choose specific columns
print("1. SELECT")
df.select("name", "salary").show()

# 2. FILTER: Select employees with salary > 50000
print("2. FILTER")
df.filter(F.col("salary") > 50000).show()

# 3. WHERE: Alternative to filter()
print("3. WHERE")
df.where(F.col("age") >= 28).show()

# 4. WITHCOLUMN: Add a bonus column (10% salary)
print("4. WITHCOLUMN")
df.withColumn(
    "bonus", F.col("salary") * 0.10
).show()

# 5. WITHCOLUMNRENAMED: Rename salary column
print("5. WITHCOLUMNRENAMED")
df.withColumnRenamed(
    "salary", "monthly_salary"
).show()

# 6. DROP: Remove age column
print("6. DROP")
df.drop("age").show()

# 7. DISTINCT: Find unique departments
print("7. DISTINCT")
df.select("department").distinct().show()

# 8. DROPDUPLICATES: Remove duplicates by name
print("8. DROPDUPLICATES")
df.dropDuplicates(["name"]).show()

# 9. ORDERBY: Sort by salary descending
print("9. ORDERBY")
df.orderBy(F.col("salary").desc()).show()

# 10. SORT: Sort by age ascending
print("10. SORT")
df.sort(F.col("age").asc_nulls_last()).show()

# 11. GROUPBY + AGG: Total salary by department
print("11. GROUPBY + AGG")
df.groupBy("department").agg(
    F.sum("salary").alias("total_salary")
).show()

# 12. JOIN: Combine employees with department locations
print("12. JOIN")
df.join(
    dept_df, on="department", how="inner"
).show()

# 13. UNION: Combine two DataFrames by position
print("13. UNION")
df1 = df.filter(F.col("id") <= 3)
df2 = df.filter(F.col("id") > 3)
df1.union(df2).show()

# 14. UNIONBYNAME: Combine by matching column names
print("14. UNIONBYNAME")
df1.unionByName(df2).show()

# 15. LIMIT: Return at most 3 rows
print("15. LIMIT")
df.limit(3).show()

# 16. SAMPLE: Randomly sample approximately 50% of rows
print("16. SAMPLE")
df.sample(
    withReplacement=False,
    fraction=0.5,
    seed=42
).show()

# 17. REPARTITION: Redistribute into 2 partitions
print("17. REPARTITION")
repartitioned_df = df.repartition(2)
repartitioned_df.show()

# 18. COALESCE: Reduce the number of partitions
print("18. COALESCE")
coalesced_df = repartitioned_df.coalesce(1)
coalesced_df.show()

# 19. INTERSECT: Find common rows
print("19. INTERSECT")
left_df = df.filter(F.col("id") <= 4)
right_df = df.filter(F.col("id") >= 3)
left_df.intersect(right_df).show()

# 20. EXCEPTALL: Find rows in left not matched in right
print("20. EXCEPTALL")
left_df.exceptAll(right_df).show()

# 21. FILLNA: Replace missing age and salary
print("21. FILLNA")
df.fillna({
    "age": 0,
    "salary": 0
}).show()

# 22. DROPNA: Remove rows with missing values
print("22. DROPNA")
df.dropna(
    subset=["age", "salary"]
).show()

# 23. REPLACE: Replace department values
print("23. REPLACE")
df.replace(
    {"IT": "Technology", "HR": "Human Resources"},
    subset=["department"]
).show()

# 24. ALIAS: Give DataFrame a temporary alias
print("24. ALIAS")
df.alias("emp").select(
    F.col("emp.name"),
    F.col("emp.salary")
).show()

# 25. SELECTEXPR: Use SQL expressions
print("25. SELECTEXPR")
df.selectExpr(
    "name",
    "salary",
    "salary * 12 AS annual_salary",
    "age + 1 AS next_year_age"
).show()

print("ALL 25 TRANSFORMATIONS COMPLETED")
```

```python
from pyspark.sql import SparkSession
from pyspark.sql.functions import *
from pyspark.sql.window import Window


# Create Retail Sales DataFrame

data = [
    (101, "Laptop",     "Electronics", "Bangalore", 75000, "2026-01-01"),
    (102, "Mobile",     "Electronics", "Bangalore", 45000, "2026-01-02"),
    (103, "Tablet",     "Electronics", "Bangalore", 45000, "2026-01-03"),
    (104, "Headphones", "Electronics", "Bangalore", 15000, "2026-01-04"),
    (105, "Camera",     "Electronics", "Bangalore", 35000, "2026-01-05"),
    (106, "Shirt",      "Clothing",    "Hyderabad",  3000, "2026-01-01"),
    (107, "Jeans",      "Clothing",    "Hyderabad",  5000, "2026-01-02"),
    (108, "Jacket",     "Clothing",    "Hyderabad",  8000, "2026-01-03"),
    (109, "T-Shirt",    "Clothing",    "Hyderabad",  3000, "2026-01-04"),
    (110, "Shoes",      "Clothing",    "Hyderabad",  6000, "2026-01-05")
]

columns = [
    "order_id",
    "product",
    "category",
    "city",
    "sales",
    "order_date"
]

df = spark.createDataFrame(data, columns)

df = df.withColumn("order_date", to_date("order_date"))

# df.show(truncate=False)
df.show()


# Define Window Specifications

# Category-wise sales ranking
w1 = Window.partitionBy("category").orderBy(col("sales").desc())

# Category-wise chronological ordering
w2 = Window.partitionBy("category").orderBy("order_date", "order_id")

# Deterministic ordering for row_number and ntile
w3 = Window.partitionBy("category").orderBy(
    col("sales").desc(), col("order_id")
)

# Complete partition window
w4 = w2.rowsBetween(
    Window.unboundedPreceding,
    Window.unboundedFollowing
)


# ==================================================
# 1. ROW_NUMBER()
# ==================================================

# Assign a unique sequential number to each product
# based on sales within each category.

df.withColumn("row_number", row_number().over(w3)).show()


# ==================================================
# 2. RANK()
# ==================================================

# Products with equal sales receive the same rank.
# The subsequent rank will contain a gap.
# Example: 1, 2, 3, 3, 5

df.withColumn("rank", rank().over(w1)).show()


# ==================================================
# 3. DENSE_RANK()
# ==================================================

# Products with equal sales receive the same rank.
# The subsequent rank will not contain a gap.
# Example: 1, 2, 3, 3, 4

df.withColumn("dense_rank", dense_rank().over(w1)).show()


# ==================================================
# 4. PERCENT_RANK()
# ==================================================

# Calculates the relative rank between 0 and 1.
# Formula: (rank - 1) / (total_rows - 1)
# Highest sales receives 0 with descending ordering.

df.withColumn("percent_rank", percent_rank().over(w1)).show()


# ==================================================
# 5. CUME_DIST()
# ==================================================

# Calculates the cumulative proportion of rows
# up to and including the current row and its ties.

df.withColumn("cumulative_distribution", cume_dist().over(w1)).show()


# ==================================================
# 6. NTILE()
# ==================================================

# Divides products into 3 sales performance groups.
# Group 1: Higher sales
# Group 2: Medium sales
# Group 3: Lower sales

df.withColumn("sales_group", ntile(3).over(w3)).show()


# ==================================================
# 7. LAG()
# ==================================================

# Retrieves the previous day's sales within a category.
# Useful for comparing current and previous sales.

df.withColumn("previous_sales", lag("sales", 1).over(w2)).show()

# Calculate sales difference

df.withColumn("previous_sales", lag("sales", 1).over(w2)) \
  .withColumn("sales_difference", col("sales") - col("previous_sales")) \
  .show()


# ==================================================
# 8. LEAD()
# ==================================================

# Retrieves the next day's sales within a category.
# Useful for comparing consecutive sales records.

df.withColumn("next_sales", lead("sales", 1).over(w2)).show()

# Calculate next sales difference

df.withColumn("next_sales", lead("sales", 1).over(w2)) \
  .withColumn("sales_difference", col("next_sales") - col("sales")) \
  .show()


# ==================================================
# 9. FIRST_VALUE()
# ==================================================

# Retrieves the first recorded sales value
# within each product category.

df.withColumn("first_sales", first_value("sales").over(w4)).show()


# ==================================================
# 10. LAST_VALUE()
# ==================================================

# Retrieves the last recorded sales value
# within each product category.
# The complete window frame is important here.

df.withColumn("last_sales", last_value("sales").over(w4)).show()


# ==================================================
# Apply All 10 Analytical Functions Together
# ==================================================

result = df \
    .withColumn("row_number", row_number().over(w3)) \
    .withColumn("rank", rank().over(w1)) \
    .withColumn("dense_rank", dense_rank().over(w1)) \
    .withColumn("percent_rank", percent_rank().over(w1)) \
    .withColumn("cume_dist", cume_dist().over(w1)) \
    .withColumn("ntile", ntile(3).over(w3)) \
    .withColumn("previous_sales", lag("sales", 1).over(w2)) \
    .withColumn("next_sales", lead("sales", 1).over(w2)) \
    .withColumn("first_sales", first_value("sales").over(w4)) \
    .withColumn("last_sales", last_value("sales").over(w4))

result.orderBy("category", col("sales").desc(), "order_id").show(truncate=False)
```
